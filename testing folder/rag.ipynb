```python
query4 = "Mijn verhuurder reageert niet, mag ik de werken in zijn plaats uitvoeren in Wallonië?"
query5 = "Ik ben student, kan het OCMW mij verplichten om een studentenjob te zoeken?"
query6 = "Ik ben werkloos. Heb ik recht op moederschapsverlof als ik werkloos ben?"
query7 = "Heeft een weeskind kind recht op een toeslag op het groeipakket in Wallonië?"
```

RAG

```python
from rags import RAG
rag = RAG()
```

SimpleRAG

```python
ids_sim,answer_sim = rag.SimpleRAG(query7)
print(f'ids: {ids_sim}')
print(f'answer: {answer_sim}')
```

ParaRAG

```python
ids_para,answer_para = rag.ParaRAG(query1)
```

```python
print(f'ids: {ids_para}')
print(f'answer: {answer_para}')
```

DecomRAG

```python
ids_decom, answer_decom = rag.DecomRAG(query1)
print(f'ids: {ids_decom}')
print(f'answer: {answer_decom}')
```
