Faithfulness

```python
from scipy import stats
from statsmodels.stats.multitest import multipletests
metric = 'faithfulness_score'
baseline = sim_rag[metric].values
results1 = []

for name, df in [('para_rag', para_rag), ('decom_rag', decom_rag), ('stepb_rag', stepb_rag), ('master_rag', master_rag), ('judge_rag', judge_rag)]:
    
    scores = df[metric].values
    
    # Wilcoxon signed-rank test (paired, non-parametric — appropriate given the
    # left-skewed, ceiling-inflated distributions seen in the violin plot)
    stat, p = stats.wilcoxon(scores, baseline, zero_method='wilcox', correction=True)
    
    # Effect size: matched-pairs rank-biserial correlation
    diffs = scores - baseline
    n_nonzero = (diffs != 0).sum()
    r_rb = stat / (n_nonzero * (n_nonzero + 1) / 2) if n_nonzero > 0 else float('nan')
    r_rb = 1 - (2 * stat) / (n_nonzero * (n_nonzero + 1) / 2) if n_nonzero > 0 else float('nan')
    
    results1.append({
        'architecture': name,
        'mean_diff': scores.mean() - baseline.mean(),
        'median_diff': (diffs).mean(),  # or np.median(diffs)
        'p_raw': p,
        'rank_biserial_r': r_rb
    })
# Holm-Bonferroni correction across the 5 comparisons vs. the shared control
pvals1 = [r['p_raw'] for r in results1]
reject1, p_adj1, _, _ = multipletests(pvals1, alpha=0.05, method='holm')
for r, p_corrected, sig in zip(results1, p_adj1, reject1):
    print(f"{r['architecture']:12s} vs master_rag: "
          f"mean diff={r['mean_diff']:+.4f}, "
          f"r={r['rank_biserial_r']:+.4f}, "
          f"p_raw={r['p_raw']:.4f}, p_holm={p_corrected:.4f}, "
          f"significant={sig}")
```

```python
from scipy import stats

def mannwhitney_analysis(df, query_col='query_type', score_col='score',
                          group1='explicit', group2='implicit', label=None):
    if label:
        print(f"=== {label} ===")

    scores1 = df[df[query_col] == group1][score_col]
    scores2 = df[df[query_col] == group2][score_col]

    n1, n2 = len(scores1), len(scores2)

    u_stat, p_value = stats.mannwhitneyu(scores1, scores2, alternative='two-sided')
    print(f"U: {u_stat}, p-value: {p_value:.4f}")

    r = 1 - (2 * u_stat) / (n1 * n2)
    print(f"Rank-biserial r ({group1} vs {group2}): {r:.4f}")

    print(f"Means: {group1}={scores1.mean():.4f} (n={n1}), {group2}={scores2.mean():.4f} (n={n2})")
    print()

    return {
        'u_stat': u_stat,
        'p_value': p_value,
        'n1': n1,
        'n2': n2,
        'rank_biserial_r': r
    }
```

```python
import re
import pandas as pd

def parse_id_string(id_str):
    if pd.isna(id_str):
        return []
    if isinstance(id_str, (list, tuple)):
        return [int(x) for x in id_str]

    # strip brackets/parentheses
    cleaned = str(id_str).strip().strip("[]()")
    # split on any whitespace (handles multiple/irregular spaces)
    tokens = re.split(r"\s+", cleaned.strip())
    tokens = [t for t in tokens if t]  # drop empty strings
    return [int(t) for t in tokens]
sim_rag["retrieved_ids"] = sim_rag["retrieved_ids"].apply(parse_id_string)
sim_rag["gold_ids"] = sim_rag["gold_ids"].apply(parse_id_string)
```

```python
import pandas as pd

def calculate_retrieval_metrics(retrieved_ids, gold_ids, k=10, total_documents=None):
    """
    Calculate retrieval metrics for a single question, following the
    Accuracy/Hit@K, Recall@K, Precision@K, and F1 definitions.

    Parameters:
        retrieved_ids (list): Ranked list of IDs returned by the retriever
                               (order matters — top result first)
        gold_ids (list): Ground-truth relevant document IDs (RD)
        k (int, optional): Cutoff for top-k retrieved documents (Top_kd).
                            If None, uses the full retrieved_ids list.
        total_documents (int, optional): Total number of documents in the
                            corpus, required only if you want Accuracy
                            (TP + TN) / TotalNumber. If None, Accuracy
                            is not computed.

    Returns:
        dict: precision@k, recall@k, f1@k, hit@k, and accuracy (if total_documents given)
    """
    # Apply top-k cutoff (Top_kd)
    top_kd = retrieved_ids[:k]
    top_kd_set = set(top_kd)
    gold_set = set(gold_ids[:k])  # RD

    # TP = relevant docs retrieved within top-k
    tp = len(top_kd_set & gold_set)
    # FP = irrelevant docs retrieved within top-k
    fp = len(top_kd_set - gold_set)

    # Recall@K = |RD ∩ Top_kd| / |RD|
    recall = tp / len(gold_set) if gold_set else 0.0
    # Precision@K = TP / (TP + FP)
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    # F1 = 2 * Precision * Recall / (Precision + Recall)
    f1 = (2 * precision * recall / (precision + recall)) if (precision + recall) > 0 else 0.0
    # Hit@K = at least one relevant doc found in top-k
    hit = 1.0 if tp > 0 else 0.0

    result = {
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "hit": hit
    }

    # Accuracy = (TP + TN) / TotalNumber, only if total_documents is provided
    if total_documents is not None:
        tn = total_documents - len(top_kd_set | gold_set)  # docs correctly NOT retrieved
        accuracy = (tp + tn) / total_documents
        result["accuracy"] = accuracy

    return result


def calculate_dataset_metrics(df, retrieved_col="retrieved_ids", gold_col="gold_ids",
                               k=10, total_documents=None):
    """
    Calculate aggregated retrieval metrics across an entire dataset.

    Parameters:
        df (pd.DataFrame): DataFrame containing retrieved and gold ID columns
                            (each cell should be a ranked list of IDs)
        retrieved_col (str): column name for retrieved IDs
        gold_col (str): column name for gold IDs
        k (int, optional): top-k cutoff to apply (e.g., 5, 10, 20)
        total_documents (int, optional): total corpus size, needed for Accuracy

    Returns:
        dict: averaged metrics across all rows
        pd.DataFrame: per-question metrics (for inspection/debugging)
    """
    results = []

    for _, row in df.iterrows():
        metrics = calculate_retrieval_metrics(
            row[retrieved_col], row[gold_col], k=k, total_documents=total_documents
        )
        results.append(metrics)

    results_df = pd.DataFrame(results)

    aggregated = {
        f"avg_precision@{k}": results_df["precision"].mean(),
        f"avg_recall@{k}": results_df["recall"].mean(),
        f"avg_f1@{k}": results_df["f1"].mean(),
        f"hit_rate@{k}": results_df["hit"].mean(),
    }

    if "accuracy" in results_df.columns:
        aggregated[f"avg_accuracy@{k}"] = results_df["accuracy"].mean()

    return aggregated
```

```python
import pandas as pd

def reciprocal_rank(retrieved_ids, gold_ids):
    """
    Calculate the reciprocal rank (1 / rank_i) for a single query.

    Parameters:
        retrieved_ids (list): Ranked list of retrieved document IDs (order matters!)
        gold_ids (list): Ground-truth relevant document IDs

    Returns:
        float: 1 / rank_i if a relevant document is found, else 0.0
               (rank_i = position of the first relevant document, 1-indexed)
    """
    gold_set = set(gold_ids)

    for rank_i, doc_id in enumerate(retrieved_ids, start=1):
        if doc_id in gold_set:
            return 1.0 / rank_i

    return 0.0  # no relevant document retrieved -> contributes 0 to the sum


def calculate_mrr(df, retrieved_col="retrieved_ids", gold_col="gold_ids"):
    """
    Calculate Mean Reciprocal Rank (MRR) across a set of queries Q:

        MRR = (1 / |Q|) * sum_{i=1}^{|Q|} (1 / rank_i)

    Parameters:
        df (pd.DataFrame): DataFrame with retrieved and gold ID columns,
                            one row per query
        retrieved_col (str): column name for ranked retrieved IDs
        gold_col (str): column name for gold IDs

    Returns:
        float: MRR score
        pd.Series: per-query reciprocal ranks (1 / rank_i), for inspection
    """
    # |Q| = number of queries
    Q = len(df)

    # 1 / rank_i for each query i = 1, ..., |Q|
    reciprocal_ranks = df.apply(
        lambda row: reciprocal_rank(row[retrieved_col], row[gold_col]),
        axis=1
    )

    # MRR = (1 / |Q|) * sum(1 / rank_i)
    mrr = (1 / Q) * reciprocal_ranks.sum() if Q > 0 else 0.0

    return mrr
```

```python
import math
import pandas as pd

def dcg_at_k(retrieved_ids, gold_ids, k=10):
    """
    Calculate Discounted Cumulative Gain at rank k:

        DCG@k = sum_{i=1}^{k} (2^rel_i - 1) / log2(i + 1)

    Parameters:
        retrieved_ids (list): Ranked list of retrieved document IDs (order matters!)
        gold_ids (list): Ground-truth relevant document IDs
        k (int): Rank cutoff

    Returns:
        float: DCG@k
    """
    gold_set = set(gold_ids)
    top_k = retrieved_ids[:k]

    dcg = 0.0
    for i, doc_id in enumerate(top_k, start=1):
        # Binary relevance: rel_i = 1 if doc_id is a gold document, else 0
        rel_i = 1 if doc_id in gold_set else 0
        dcg += (2 ** rel_i - 1) / math.log2(i + 1)

    return dcg


def idcg_at_k(gold_ids, k=10):
    """
    Calculate Ideal DCG at rank k — the maximum possible DCG@k,
    obtained by ranking all relevant documents first.

    Parameters:
        gold_ids (list): Ground-truth relevant document IDs
        k (int): Rank cutoff

    Returns:
        float: IDCG@k
    """
    # Ideal ranking: all relevant docs (rel_i = 1) placed at the top,
    # up to min(|gold_ids|, k) positions
    num_relevant = min(len(set(gold_ids)), k)

    idcg = 0.0
    for i in range(1, num_relevant + 1):
        rel_i = 1
        idcg += (2 ** rel_i - 1) / math.log2(i + 1)

    return idcg


def ndcg_at_k(retrieved_ids, gold_ids, k=10):
    """
    Calculate Normalized Discounted Cumulative Gain at rank k:

        NDCG@k = DCG@k / IDCG@k

    Parameters:
        retrieved_ids (list): Ranked list of retrieved document IDs
        gold_ids (list): Ground-truth relevant document IDs
        k (int): Rank cutoff

    Returns:
        float: NDCG@k (0.0 if IDCG@k is 0, e.g. no gold documents)
    """
    dcg = dcg_at_k(retrieved_ids, gold_ids, k)
    idcg = idcg_at_k(gold_ids, k)

    return dcg / idcg if idcg > 0 else 0.0


def calculate_dataset_ndcg(df, retrieved_col="retrieved_ids", gold_col="gold_ids", k=10):
    """
    Calculate average NDCG@k across an entire dataset.

    Parameters:
        df (pd.DataFrame): DataFrame with retrieved and gold ID columns
        retrieved_col (str): column name for ranked retrieved IDs
        gold_col (str): column name for gold IDs
        k (int): Rank cutoff

    Returns:
        float: average NDCG@k across all queries
        pd.Series: per-query NDCG@k scores (for inspection)
    """
    ndcg_scores = df.apply(
        lambda row: ndcg_at_k(row[retrieved_col], row[gold_col], k=k),
        axis=1
    )
    avg_ndcg = ndcg_scores.mean()

    return avg_ndcg
```

```python
sim_avg_ndcg10 = calculate_dataset_ndcg(sim_rag, retrieved_col="retrieved_ids", gold_col="gold_ids", k=10)
print(f"NDCG@10 simpleRAG: {sim_avg_ndcg10:.4f}")
para_avg_ndcg10 = calculate_dataset_ndcg(para_rag, retrieved_col="retrieved_ids", gold_col="gold_ids", k=10)
print(f"NDCG@10 paraRAG: {para_avg_ndcg10:.4f}")
decom_avg_ndcg10 = calculate_dataset_ndcg(decom_rag, retrieved_col="retrieved_ids", gold_col="gold_ids", k=10)
print(f"NDCG@10 decomRAG: {decom_avg_ndcg10:.4f}")
stepb_avg_ndcg10 = calculate_dataset_ndcg(stepb_rag, retrieved_col="retrieved_ids", gold_col="gold_ids", k=10)
print(f"NDCG@10 stepbRAG: {stepb_avg_ndcg10:.4f}")
master_avg_ndcg10 = calculate_dataset_ndcg(master_rag, retrieved_col="retrieved_ids", gold_col="gold_ids", k=10)
print(f"NDCG@10 masterRAG: {master_avg_ndcg10:.4f}")
judge_avg_ndcg10 = calculate_dataset_ndcg(judge_rag, retrieved_col="retrieved_ids", gold_col="gold_ids", k=10)
print(f"NDCG@10 judgeRAG: {judge_avg_ndcg10:.4f}")
```
